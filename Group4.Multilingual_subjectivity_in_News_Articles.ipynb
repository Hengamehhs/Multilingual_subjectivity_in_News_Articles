```python
# Preprocessing
import re

def clean_text(text):
  text = text.lower()  # Convert to lowercase
  text = re.sub(r"[^a-z0-9\s]", "", text)  # Remove non-alphanumeric characters and spaces
  text = re.sub(r"\s+", " ", text)  # Remove extra spaces
  return text

# Apply the cleaning function to each sentence
df['cleaned_sentence'] = df['sentence'].apply(clean_text)
```

```python
from transformers import BertTokenizer

tokenizer = BertTokenizer.from_pretrained('bert-base-uncased', do_lower_case=True)
```

```python
#use preprocessed sentences
sentences = df['cleaned_sentence'].values
```

```python
# encode all tokenized the sentences
input_ids = []

for sentence in sentences:
    encoded_sentence = tokenizer.encode(sentence)
    input_ids.append(encoded_sentence)

# example
print(sentences[0])
print(input_ids[0])
```

```python
# We will use some utility function from tensorflow
from keras.preprocessing.sequence import pad_sequences

MAX_LEN = 150

#Padding the input to the max length that is 64
input_ids = pad_sequences(input_ids, maxlen=MAX_LEN, dtype="long", value=0, truncating="post", padding="post")

# example
print(sentences[0])
print(input_ids[0])
```

```python
# Creating the attention masks
attention_masks = []
for i in input_ids:
    att_mask = [int(token_id > 0) for token_id in i]
    attention_masks.append(att_mask)

print(attention_masks[0])
print(attention_masks[1])
```

```python
from transformers import BertForSequenceClassification, AdamW, BertConfig

model = BertForSequenceClassification.from_pretrained(
    "bert-base-uncased",
    num_labels = 2,
    output_attentions = False,
    output_hidden_states = False,)
```

```python
# Convert labels to binary values of 0 and 1
label_mapping = {'OBJ': 0, 'SUBJ': 1}

# Use map function to replace values in the "label" column
df['labelBinary'] = df['label'].map(label_mapping)

labels = df['labelBinary'].values
print(labels)
```

```python
import torch
from torch.utils.data import TensorDataset, DataLoader, RandomSampler, SequentialSampler

# Create a TensorDataset
dataset = TensorDataset(torch.tensor(input_ids), torch.tensor(attention_masks), torch.tensor(labels))

# Create a DataLoader
dataloader = DataLoader(dataset, batch_size=8, shuffle=True)
```

```python
print(torch.tensor(input_ids).shape)
print(torch.tensor(attention_masks).shape)
print(torch.tensor(labels).shape)
```

```python
# Set up training parameters
optimizer = AdamW(model.parameters(), lr=5e-5)
device = torch.device('cuda') if torch.cuda.is_available() else torch.device('cpu')
model.to(device)
```

```python
# Train the model
model.train()
for epoch in range(2):  # Adjust the number of epochs as needed
    for batch in dataloader:
        input_ids = batch[0].to(device)
        attention_mask = batch[1].to(device)
        labels = batch[2].to(device)

        optimizer.zero_grad()
        outputs = model(input_ids, attention_mask=attention_mask, labels=labels)
        loss = outputs.loss
        loss.backward()
        optimizer.step()

    print(f"Epoch {epoch + 1}, Loss: {loss.item()}")
```

```python
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
import re
import pandas as pd
import requests
import json
```

```python

# Define the cleaning function
def clean_text(text):
    text = re.sub(r'\W', ' ', str(text))
    text = text.lower()
    text = re.sub(r'\s+[a-zA-Z]\s+', ' ', text)
    lemmatizer = WordNetLemmatizer()
    tokens = text.split()
    tokens = [lemmatizer.lemmatize(word) for word in tokens if word not in stopwords.words('english')]
    return ' '.join(tokens)

# Read the dataset
file_path = "C:\\Users\\hngho\\DAT550\\task2\\data\\subtask-2-english\\dev_test_en.tsv"
df = pd.read_csv(file_path, delimiter='\t')
df['cleaned_sentence'] = df['sentence'].apply(clean_text)
# cleaned_sentence column for prediction
cleaned_sentences = df['cleaned_sentence'].values
#print(cleaned_sentences)
selected_sentences = cleaned_sentences[:30]
print(len(cleaned_sentences))

```

```python
import requests
import json
import re

def check_subjectivity(sentence):
    url = "http://localhost:11434/api/generate"  
    headers = {'Content-Type': 'application/json'}
    prompt = f"Determine if the following statement is subjective or objective:\n\n'{sentence}'\n\nResponse: "
    data = {
        "model": "mistral", 
        "prompt": prompt
    }
    response = requests.post(url, headers=headers, json=data)
    if response.status_code == 200:
        try:
            for line in response.text.split('\n'):
                if line.strip():
                    response_data = json.loads(line)
                    result = response_data.get('response', '').strip()
                    # Check if response have subjectivity or objectivity senteces
                    if re.search(r'\b(subjective|objective)\b', result, re.IGNORECASE):
                        return result
             # return 'Unknown,if was not subjective or objective
            return 'Unknown'
        except json.JSONDecodeError as e:
            return f"JSON decoding failed: {str(e)}"
    else:
        return f"Failed to get response: {response.status_code}"
for sentence in selected_sentences:
    result = check_subjectivity(sentence)
    print(f"Sentence: {sentence}\nLabel: {result}\n")
```

## Zero shot classification 

```python
import requests

def check_subjectivity(sentence):
    url = "http://localhost:11434/api/generate"
    headers = {'Content-Type': 'application/json'}
    prompt = f"Is the following sentence subjective or objective? Please respond with 'subjective' or 'objective' only. Sentence: '{sentence}'"
    data = {
        "model": "mistral",
        "prompt": prompt
    }
    response = requests.post(url, headers=headers, json=data, stream=True)
    full_response = ''
    try:
        for line in response.iter_lines():
            if line:
                decoded_line = line.decode('utf-8')
                json_line = json.loads(decoded_line)
                full_response += json_line['response']
                if json_line['done']:
                    break
        return full_response.strip()
    except json.JSONDecodeError as e:
        return f"JSON decoding failed: {str(e)}"

sentence = "The sky is incredibly beautiful today."
result = check_subjectivity(sentence)
print(result)
```

## Few shot calssification 

```python

def check_subjectivity(sentence):
    url = "http://localhost:11434/api/generate"  
    headers = {'Content-Type': 'application/json'}
    prompt = f"""Determine if the following statement is subjective or objective:
    Here are a few examples of subjectivity prompts:
    - History shows that McCarthy and McConnell, like so many Republican leaders, collapse after demands from the left. (label: objective)
    - The sky is incredibly beautiful today. (subjective)
    
    '{sentence}'
    
    Response: """
    data = {
        "model": "mistral",  
        "prompt": prompt
    }
    response = requests.post(url, headers=headers, json=data)
    if response.status_code == 200:
        try:
            for line in response.text.split('\n'):
                if line.strip():
                    response_data = json.loads(line)
                    result = response_data.get('response', '').strip()
                    return result
        except json.JSONDecodeError as e:
            return f"JSON decoding failed: {str(e)}"
    else:
        return f"Failed to get response: {response.status_code}"
    

sentence = "Doctors recommend at least thirty minutes of physical activity every day to maintain a healthy lifestyle."
result = check_subjectivity(sentence)
print(result)
```
